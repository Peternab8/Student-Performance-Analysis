#Student Performance Analysis


## 🧠 About Dataset

### 📄 Description

This dataset offers a comprehensive overview of factors influencing student academic performance. It captures a range of variables including study habits, attendance, parental involvement, access to resources, and lifestyle factors — all contributing to exam outcomes.

---

### 📊 Column Descriptions

| **Attribute**                  | **Description**                                                                 |
| ------------------------------ | ------------------------------------------------------------------------------- |
| **Hours_Studied**              | Number of hours spent studying per week.                                        |
| **Attendance**                 | Percentage of classes attended.                                                 |
| **Parental_Involvement**       | Level of parental involvement in the student's education *(Low, Medium, High)*. |
| **Access_to_Resources**        | Availability of educational resources *(Low, Medium, High)*.                    |
| **Extracurricular_Activities** | Participation in extracurricular activities *(Yes, No)*.                        |
| **Sleep_Hours**                | Average number of hours of sleep per night.                                     |
| **Previous_Scores**            | Scores from previous exams.                                                     |
| **Motivation_Level**           | Student's level of motivation *(Low, Medium, High)*.                            |
| **Internet_Access**            | Availability of internet access *(Yes, No)*.                                    |
| **Tutoring_Sessions**          | Number of tutoring sessions attended per month.                                 |
| **Family_Income**              | Family income level *(Low, Medium, High)*.                                      |
| **Teacher_Quality**            | Quality of the teachers *(Low, Medium, High)*.                                  |
| **School_Type**                | Type of school attended *(Public, Private)*.                                    |
| **Peer_Influence**             | Influence of peers on academic performance *(Positive, Neutral, Negative)*.     |
| **Physical_Activity**          | Average number of hours of physical activity per week.                          |
| **Learning_Disabilities**      | Presence of learning disabilities *(Yes, No)*.                                  |
| **Parental_Education_Level**   | Highest education level of parents *(High School, College, Postgraduate)*.      |
| **Distance_from_Home**         | Distance from home to school *(Near, Moderate, Far)*.                           |
| **Gender**                     | Gender of the student *(Male, Female)*.                                         |
| **Exam_Score**                 | Final exam score.                                                               |

# 1️⃣ Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# 2️⃣ Load Dataset

```python
df = pd.read_csv('/content/StudentPerformanceFactors.csv')
```

# 4️⃣ Data Preprocessing & Cleaning


```python
df.columns = df.columns.str.strip().str.replace(' ', '_')
```

```python
print("Columns:", df.columns.tolist())
```

```python
df.drop_duplicates(inplace=True)
```

```python
cat_cols = df.select_dtypes('object').columns
df[cat_cols] = df[cat_cols].apply(lambda x: x.str.strip().str.lower())
```

```python
df.dropna(inplace=True)
```

```python
print("Missing Values after dropping:", df.isnull().sum().sum())
```

```python
print("Info after dropping:", df.info())
```

```python
print("Summary Stats after dropping:",df.describe() )
```

```python
print("Data cleaned successfully")
```

#5️⃣ Correlation & Insights

```python
num_df = df.select_dtypes(include=[np.number])

plt.figure(figsize=(10,6))
sns.heatmap(num_df.corr(), annot=True, cmap='YlGnBu')
plt.title('Correlation Heatmap')
plt.show()
```

```python
motivation_counts = df['Motivation_Level'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    motivation_counts,
    labels=motivation_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=['#66b3ff', '#99ff99', '#ffcc99']
)

plt.title('Distribution of Students by Motivation Level', fontsize=14)
plt.show()
```

#6️⃣ Visualization

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x='Hours_Studied',
    y='Exam_Score',
    hue='Motivation_Level',
    palette='coolwarm',
    s=80
)
plt.title('Relationship Between Study Hours and Exam Score', fontsize=14)
plt.xlabel('Hours Studied per Week')
plt.ylabel('Final Exam Score')
plt.legend(title='Motivation Level')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Hours_Studied', y='Exam_Score', hue='Parental_Involvement')
plt.title('Exam Score vs Hours Studied (colored by Parental Involvement)')
plt.show()
```
